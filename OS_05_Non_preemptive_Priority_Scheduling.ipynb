```python
#Non pre emptive Priority Queue -- Lower value, Higher Priority
processes = [
  {"PID": "P1", "AT": 0, "BT": 3, "PR": 4},
  {"PID": "P2", "AT": 2, "BT": 2, "PR": 3},
  {"PID": "P3", "AT": 1, "BT": 1, "PR": 1},
  {"PID": "P4", "AT": 4, "BT": 2, "PR": 0},
  {"PID": "P5", "AT": 3, "BT": 2, "PR": 2},
  {"PID": "P6", "AT": 5, "BT": 3, "PR": 2}
  ]

time = 0
completed = []
total_wt = 0
total_tat = 0
#=====================================================================================================================
def find_highest_priority(ready):
  highest = ready[0]

  for p in ready:
    if p["PR"] < highest["PR"]:
      highest = p
  return highest


while len(completed) < len(processes):
  ready = []
  for p in processes:
    if p["AT"] <= time and p not in completed:
      ready.append(p)

  if not ready:
    time += 1
    continue
  p = find_highest_priority(ready)

#==========================
  time += p["BT"]

  p["CT"] = time
  p["TAT"] = p["CT"] - p["AT"]
  p["WT"] = p["TAT"] - p["BT"]

  total_wt += p["WT"]
  total_tat += p["TAT"]
  completed.append(p)


print("PID\tAT\tBT\tCT\tTAT\tWT")
for p in completed:
  print(
    p["PID"], "\t",
    p["AT"], "\t",
    p["BT"], "\t",
    p["CT"], "\t",
    p["TAT"], "\t",
    p["WT"]
    )


# Average Waiting Time
average_wt = total_wt / len(processes)
# Average Turnaround Time
average_tat = total_tat / len(processes)

print("\nAverage Waiting Time =", average_wt)
print("Average Turnaround Time =", average_tat)
```
