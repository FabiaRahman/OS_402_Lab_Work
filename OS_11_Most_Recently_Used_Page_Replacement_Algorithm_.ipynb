# MRU Page Replacement Algorithm

```python
pages = [1,2,3,4,1,2,3,1]
frames = []
size = 3

fault = 0
hit = 0
recent = None

for p in pages:
  if p in frames:
    hit += 1
    recent = p
    print(p, frames, "Hit")

  else:
    fault +=1

    if len(frames)<size:
      frames.append(p)
    else:
     frames[frames.index(recent)] = p
    recent = p
    print(p, frames, "Fault")

print("\nFault:",fault)
print("Hit:",hit)
print("Hit Ratio:",hit/len(pages))
```
