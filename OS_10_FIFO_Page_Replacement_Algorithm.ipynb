# FIFO - Page Replacement Algorithm

```python
pages = [1,2,3,4,1,2,5,1,2,3,4,5]
frames = []
size = 3

fault = 0
hit = 0

for p in pages:
  if p in frames:
    hit += 1
  else:
    fault +=1

    if len(frames)<size:
      frames.append(pages)
    else:
     frames.pop(0)
     frames.append(p)

print("Fault:",fault)
print("Hit:",hit)
print("Hit Ratio:",hit/len(pages))
```
