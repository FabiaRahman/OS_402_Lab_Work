```python
# Round Robin fixed
processes = [
    {"PID": "P1", "AT": 0, "BT": 7},
    {"PID": "P2", "AT": 1, "BT": 4},
    {"PID": "P3", "AT": 2, "BT": 15},
    {"PID": "P4", "AT": 3, "BT": 11},
    {"PID": "P5", "AT": 4, "BT": 20},
    {"PID": "P6", "AT": 4, "BT": 9},
]

quantum = int(input("Enter Time Quantum: "))

time = 0
completed = []
queue = []
total_wt = 0
total_tat = 0

# store burst time
for p in processes:
    p["RT"] = p["BT"]

while len(completed) < len(processes):
    # add newly arrived to queue
    for p in processes:
        if (p["AT"] <= time and p not in completed
                and p not in queue and p["RT"] > 0):
            queue.append(p)
    # queue = empty, CPU= idle
    if not queue:
        time += 1
        continue
    # Take first process from queue
    p = queue.pop(0)
    # Run for quantum or remaining time
    if p["RT"] > quantum:
        time += quantum
        p["RT"] -= quantum
    else:
        time += p["RT"]
        p["RT"] = 0
        p["CT"] = time
        p["TAT"] = p["CT"] - p["AT"]
        p["WT"] = p["TAT"] - p["BT"]
        total_wt += p["WT"]
        total_tat += p["TAT"]
        completed.append(p)

    # adds newly arrived process
    for p2 in processes:
        if (p2["AT"] <= time and p2 not in completed
                and p2 not in queue and p2["RT"] > 0
                and p2 != p):
            queue.append(p2)
    # process is not finished, put it at the end
    if p["RT"] > 0:
        queue.append(p)

print("PID\tAT\tBT\tCT\tTAT\tWT")
for p in completed:
    print(
        p["PID"], "\t",
        p["AT"], "\t",
        p["BT"], "\t",
        p["CT"], "\t",
        p["TAT"], "\t",
        p["WT"]
    )

# Average Waiting Time
average_wt = total_wt / len(processes)
# Average Turnaround Time
average_tat = total_tat / len(processes)
print("\nAverage Waiting Time =", average_wt)
print("Average Turnaround Time =", average_tat)
```

```python
# SJF
processes = [
    {"PID": "P1", "AT": 0, "BT": 7},
    {"PID": "P2", "AT": 1, "BT": 4},
    {"PID": "P3", "AT": 2, "BT": 15},
    {"PID": "P4", "AT": 3, "BT": 11},
    {"PID": "P5", "AT": 4, "BT": 20},
    {"PID": "P6", "AT": 4, "BT": 9},
]

time = 0
completed = []
total_wt = 0
total_tat = 0

def find_shortest(ready):
    shortest = ready[0]

    for p in ready:
        if p["BT"] < shortest["BT"]:
            shortest = p
    return shortest

while len(completed) < len(processes):
    ready = []
    for p in processes:
        if p["AT"] <= time and p not in completed:
            ready.append(p)
    if not ready:
        time += 1
        continue
    p = find_shortest(ready)
    time += p["BT"]

    p["CT"] = time
    p["TAT"] = p["CT"] - p["AT"]
    p["WT"] = p["TAT"] - p["BT"]

    total_wt += p["WT"]
    total_tat += p["TAT"]
    completed.append(p)
print("PID\tAT\tBT\tCT\tTAT\tWT")

for p in completed:
    print(
        p["PID"], "\t",
        p["AT"], "\t",
        p["BT"], "\t",
        p["CT"], "\t",
        p["TAT"], "\t",
        p["WT"]
    )

# Average Waiting Time
average_wt = total_wt / len(processes)
# Average Turnaround Time
average_tat = total_tat / len(processes)
print("\nAverage Waiting Time =", average_wt)
print("Average Turnaround Time =", average_tat)
```

```python
# FCFS

process_list = [
    {"PID": "P1", "AT": 0, "BT": 7},
    {"PID": "P2", "AT": 1, "BT": 4},
    {"PID": "P3", "AT": 2, "BT": 15},
    {"PID": "P4", "AT": 3, "BT": 11},
    {"PID": "P5", "AT": 4, "BT": 20},
    {"PID": "P6", "AT": 4, "BT": 9},
]


def get_arrival_time(item):
    return item["AT"]


# Sort by Arrival Time
process_list.sort(key=get_arrival_time)

time = 0
total_tat = 0
total_wt = 0

print("PID\tAT\tBT\tCT\tTAT\tWT")

# Calculate CT, TAT, and WT
for item in process_list:
    PID = item["PID"]
    AT = item["AT"]
    BT = item["BT"]

    if time < AT:
        time = AT

    time += BT

    CT = time
    TAT = CT - AT
    WT = TAT - BT

    total_tat += TAT
    total_wt += WT

    print(
        PID, "\t",
        AT, "\t",
        BT, "\t",
        CT, "\t",
        TAT, "\t",
        WT
    )


# Average Turnaround Time
avg_tat = total_tat / len(process_list)

# Average Waiting Time
avg_wt = total_wt / len(process_list)

print("\nAverage Turnaround Time:", avg_tat)
print("Average Waiting Time:", avg_wt)
```
