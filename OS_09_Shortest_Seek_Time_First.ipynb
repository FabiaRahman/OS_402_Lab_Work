# SSTF

```python
req_sequence = [60,20,100,10,15,22,42]
head = 50

total_seek = 0
current = head
visited = [False] * len(req_sequence)
path = []

for i in range(len(req_sequence)):
  min_distance = float('inf')
  index = -1

  for j in range(len(req_sequence)):
    if not visited[j]:
      distance = abs(current - req_sequence[j])
      if distance < min_distance:
        min_distance = distance
        index = j

  visited[index] = True
  total_seek += min_distance
  path.append((current, req_sequence[index], min_distance))
  current = req_sequence[index]

for start, end, dist in path:
  print(start, "=>", end, "=> (Seek:", dist, ")")

print("\nTotal Seek:", total_seek)
```
