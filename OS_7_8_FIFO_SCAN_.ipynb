# First Come, First Served

```python
#unsorted
req_sequence = [0,41,30,100,62,51,20]
head = 70

total_seek = 0
current = head

for r in req_sequence:
  difference = abs(current-r)
  print(current,"=>",r,"=>",difference)
  total_seek += difference
  current = r

print("\nTotal Seek",total_seek)
```

```python
#sorted
req_sequence = [0,41,30,100,62,51,20]
req_sequence.sort()
print("Sorted sequence:", req_sequence)
head = 70

total_seek = 0
current = head

for r in req_sequence:
  difference = abs(current-r)
  print(current,"=>",r,"=>",difference)
  total_seek += difference
  current = r

print("\nTotal Seek",total_seek)
```

# SCAN

```python
#FOR LEFT

request_sequence =[0,41,30,100,62,51,20]
head = 70
total = 0

for request in [62,51,41,30,20,0]:
  movement = abs (request - head)
  total  += movement
  print ("Head:", head , "Request:", request, "Movement:", movement)
  head = request

print ("\nTotal Movement:", total)


for request in [100]:
  movement = abs (request - head)
  total  += movement
  print ("Head:", head , "Request:", request, "Movement:", movement)
  head = request

print ("\nTotal Movement:", total)
```
